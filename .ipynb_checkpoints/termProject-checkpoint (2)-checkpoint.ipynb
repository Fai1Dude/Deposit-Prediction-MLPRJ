```python
# a-ii. Make univariate graphs.
print("ii. univariate graphs.")

# For non-numeric data columns:
num_columns = df.select_dtypes(exclude='object').columns
cat_columns = df.select_dtypes(include='object').columns

%matplotlib inline
from matplotlib import pyplot as plt
import seaborn as sns

nCols=3
nRows=len(cat_columns)//nCols+1


fig,axes = plt.subplots(nRows, nCols, figsize=(25,25))
for ind,col in enumerate(cat_columns):
    sns.countplot(y=col,data=df,ax=axes.flatten()[ind])
print("For non-numeric data columns:")
plt.show()

# For numeric data columns:
print("For numeric data columns:")
for col in num_columns:
    plt.figure(figsize=(5,5))
    sns.histplot(x=col,bins=10,kde=True,data=df)
    plt.xticks(rotation=45)
    plt.show()
    
```

# 4. Model Building.



```python
#a. Perform the modeling by utilizing the selective parameters obtained from crossvalidation.

#df.head()
#df.isna().sum()
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('bank.csv', delimiter = ',')
display(df.head())
display(df.info())
#display(df["Acceptable"].unique())
job_mapping = {'unemployed':0, 
        'services':1, 
        'management':1, 
        'blue-collar':1,
        'self-employed':1,
        'technician':1,
        'entrepreneur':1,
        'admin.':1,
        'student':1,
        'housemaid':1,
        'retired':0,
        'unknown':0}
marital_mapping = {'single':0, 
        'married':1, 
        'divorced':0}
education_mapping = {'unknown':0, 
        'primary':1, 
        'secondary':2,
        'tertiary':3 }
default_mapping = {'no':1, 
        'yes':0}
#pdays_mapping = {-1:0}
housing_mapping = {'no':0, 
        'yes':1}
loan_mapping = {'no':0, 
        'yes':1}
contact_mapping = {'unknown':0, 
        'telephone':1, 
        'cellular':1}
month_mapping = {'jan':0, 
        'feb':0, 
        'mar':0, 
        'apr':0,
        'may':0,
        'jun':0,
        'jul':0,
        'aug':0,
        'sep':0,
        'oct':0,
        'nov':0,
        'dec':0}

poutcome_mapping = {'unknown':0, 
        'failure':0, 
        'other':1,
        'success':2, }
                   
y_mapping = {'no':0, 
        'yes':1}
                   
                   

ndf = df
ndf['job'] = df['job'].map(job_mapping)
ndf['marital'] = df['marital'].map(marital_mapping)
ndf['education'] = df['education'].map(education_mapping)
ndf['default'] = df['default'].map(default_mapping)
ndf['housing'] = df['housing'].map(housing_mapping)                  
ndf['loan'] = df['loan'].map(loan_mapping)
ndf['contact'] = df['contact'].map(contact_mapping)
ndf['month'] = df['month'].map(month_mapping)
ndf['poutcome'] = df['poutcome'].map(poutcome_mapping)
ndf['y'] = df['y'].map(y_mapping)
ndf['pdays'] = df['pdays'].replace({-1:0})

display(ndf.head())


from sklearn.model_selection import train_test_split
X = ndf.drop('y',axis=1).values
y = ndf['y'].values 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

from sklearn import tree
dtClf = tree.DecisionTreeClassifier(random_state=0,criterion='entropy',splitter='best')
dtClf = dtClf.fit(X_train,y_train)
dt_y_pred = dtClf.predict(X_test)


import matplotlib.pyplot as plt
plt.figure(figsize =(8,8),dpi=1000)
tree.plot_tree(dtClf,feature_names=df.columns[:-1].tolist(),class_names=['0','1'],
               filled=True,rounded=True,max_depth=5); #
plt.show()
```

```python
#b. Compare different models based on the performance measures.

from sklearn.metrics import accuracy_score
print("Decision Tree: \n")
print("Accuracy:=",  accuracy_score(y_test, dt_y_pred))




# Generate Train - Test splits
from sklearn.model_selection import train_test_split
X = df.iloc[:,:-1].values
y = df.iloc[:, -1].values 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


# Scaling the Train - Test splits
import numpy as np
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scaler.fit(np.c_[X_train,y_train])

A_train = scaler.transform(np.c_[X_train,y_train])
X_train = A_train[:,:-1]
y_train = A_train[:,-1]

A_test = scaler.transform(np.c_[X_test,y_test])
X_test = A_test[:,:-1]
y_test = A_test[:,-1]

print()
print()

print("regression:")

# Regression Analysis: Mean Squared Error Metric
from sklearn.metrics import mean_squared_error

## OLS
from sklearn.linear_model import LinearRegression
reg1 = LinearRegression(fit_intercept=False).fit(X_train, y_train)
y_pred1 = reg1.predict(X_test)
print('The MSE using OLS is:', mean_squared_error(y_test, y_pred1))


from sklearn.linear_model import RidgeCV
reg2 = RidgeCV(alphas=[1e-3, 1e-2, 1e-1, 1e0, 1e1, 1e2, 1e3], fit_intercept=False,cv=10).fit(X_train, y_train)
y_pred2 = reg2.predict(X_test)
print('The MSE using Ridge is:', mean_squared_error(y_test, y_pred2))


## Lasso
from sklearn.linear_model import LassoCV
reg3 = LassoCV(alphas=[1e-3, 1e-2, 1e-1, 1e0, 1e1, 1e2, 1e3], 
               fit_intercept=False,cv=10, random_state=0).fit(X_train, y_train)
y_pred3 = reg3.predict(X_test)
print('The MSE using Lasso is:', mean_squared_error(y_test, y_pred3))


#remove this when submitting!!!!
#you have to run the cell before this for this cell to work
```

```python
### b. Compare different models based on the performance measures.
import numpy as np
from sklearn import metrics
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

kmeans = KMeans(n_clusters=3, max_iter=1000, n_init=10, init='random', random_state=0).fit(df)


k_vals=range(2,15)

dfA = df

# Empty arrays for internal measures
sShil_max = np.empty([len(k_vals),1])

for i,n in enumerate(k_vals):
    kmeans = KMeans(n_clusters=n, max_iter=1000, n_init=10, random_state=0).fit(dfA)
    sShil_max[i] = metrics.silhouette_score(dfA,kmeans.labels_)
    
    
plt.plot(k_vals, sShil_max, 'o:',c='r') #  marker|line|color fromat

plt.xlabel("K")
plt.ylabel("Cluster Indices")
plt.legend(['SI'])
plt.show()
```
