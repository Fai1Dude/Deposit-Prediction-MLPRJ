```python
#1. Read the selected data
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
df = pd.read_csv('bank.csv', delimiter = ',')
display(df.head())
```

```python
#2. list the fields/variables
df.columns
# print(df.select_dtypes("object").describe().loc['unique', :])
# df['y'].unique().tolist()
```

```python
#3. identify their types
df.dtypes
```

```python
#4. List the inconsistencies, missing data, and outliers
df.isnull().sum()
## duration, pdays have missing data
```

```python
#5. Fix the inconsistencies, impute the missing data and remove the outliers.
df=df.drop('duration',axis=1)
df=df.drop('pdays',axis=1)
```

```python
#6. show data after fixing it:
df.head()
```

# 3. Model Planing.

```python
# a-i. Find the summary statistics.

# display(df.describe(include='all'))
display(df.describe(include='number'))
display(df.describe(include='object'))
```

```python
# a-ii. Make univariate graphs.
print("ii. univariate graphs.")

# For non-numeric data columns:
num_columns = df.select_dtypes(exclude='object').columns
cat_columns = df.select_dtypes(include='object').columns

%matplotlib inline
from matplotlib import pyplot as plt
import seaborn as sns

nCols=3
nRows=len(cat_columns)//nCols+1


fig,axes = plt.subplots(nRows, nCols, figsize=(25,25))
for ind,col in enumerate(cat_columns):
    sns.countplot(y=col,data=df,ax=axes.flatten()[ind])
print("For non-numeric data columns:")
plt.show()

# For numeric data columns:
print("For numeric data columns:")
for col in num_columns:
    plt.figure(figsize=(5,5))
    sns.histplot(x=col,bins=10,kde=True,data=df)
    plt.xticks(rotation=45)
    plt.show()
    
```

```python
# a-iii. Prepare bivariate plots.
print("iii. Bivariate plots.")

# Histograms for each non-numeric data column differentiated with *y (subscribed)* column.
print("For each non-numeric data column differentiated with *y (subscribed)*")
nCols=3
nRows=len(cat_columns)//nCols+1

fig,axes = plt.subplots(nRows, nCols, figsize=(25,25))
for ind,col in enumerate(cat_columns):
    sns.countplot(y=col,hue='y',data=df,ax=axes.flatten()[ind])
plt.show()

# For every pair of numerical columns differentiated with *subscribed* column.
print("For every pair of numerical columns differentiated with *y (subscribed)* column)*")
for col in num_columns:
    plt.figure(figsize=(5,5))
    sns.histplot(x=col,bins=10,hue='y',kde=True,data=df,palette=['r','b'])
    plt.xticks(rotation=45)
    plt.show()
```

```python
# a-iv. Prptray advanced graphs
print("iv. Bivariate plots.")

#  For non-numeric data columns
print("For non-numeric data columns")

# depicting *job*, and differentiated with *subscribed* & *education* columns.
plt.figure()
g=sns.catplot(x="job", col="y", hue="education", data=df, kind="count")
g.set_xticklabels(rotation=80)
plt.show()

# depicting *job*, and differentiated with *subscribed* & *marital* columns.
plt.figure()
g=sns.catplot(x="job", col="y", hue="marital", data=df, kind="count")
g.set_xticklabels(rotation=80)
plt.show()

# depicting *marital*, and differentiated with *subscribed* & *education* columns.
plt.figure()
g=sns.catplot(x="marital", col="y", hue="education", data=df, kind="count")
g.set_xticklabels(rotation=80)
plt.show()

#  For numeric data columns
print("For numeric data columns")

plt.figure()
sns.pairplot(vars=num_columns,hue ='y',data=df)   # look for {x, y}_vars kwargs
plt.show()
```
